1. Write a Python program to read content from a file and display the word with maximum length.

```python
def max_len(li):
  max = li[0]
  for i in li:
    if len(i) > len(max):
      max = i
  return max

with open('/content/Practical-7/sample_text.txt') as file:
  output = file.read()
  max_len_string = max_len(output.split())
  print("File Content: \n" + output)
  print("\nWord With Max Length:", max_len_string)
```

1. Write a program that performs text file analysis:
- Read a text file containing student essays
- Create a new file that stores:
    - Total number of sentences
    - Average words per sentence
    - List of all words longer than 10 characters
- Output should be saved in &#39;analysis.txt&#39;

```python
with open('/content/sample_data/essays.txt') as file:
    essay = file.read()

# Remove extra whitespace
essay = essay.strip()

# Split into sentences and remove empty ones
sentences = [s.strip() for s in essay.split('.') if s.strip()]

num_of_sentences = len(sentences)

total_words = 0
word_longer_than_10_ch = []

for sentence in sentences:
    words = sentence.split()
    total_words += len(words)

    for word in words:
        clean_word = word.strip(",!?\"'")
        if len(clean_word) > 10:
            word_longer_than_10_ch.append(clean_word)

avg_words_per_sentence = total_words / num_of_sentences

analysis = f"""
Total number of sentences: {num_of_sentences}
Average words per sentence: {avg_words_per_sentence}
Words longer than 10 characters: {word_longer_than_10_ch}
"""

with open('/content/sample_data/analysis.txt', 'w') as file:
    file.write(analysis.strip())

print(analysis.strip())
print("Analysis is Saved in analysis.txt")
```


2. Create a program to handle employee records:
- Write employee details (id, name, salary) to &#39;employees.csv&#39;
- Read the CSV file and create two new files:
  - high_salary.txt (employees with salary &gt; 50000)
  - low_salary.txt (employees with salary &lt;= 50000)
- Each output file should maintain the same CSV format

```python
with open('/content/sample_data/employees.csv') as file:
  csv_format = file.readline()
  high_sal = []
  low_sal = []
  for i in file:
    emp = i.strip('\n ').split(',')
    if int(emp[2]) > 50000:
      high_sal.append(emp)
    else:
      low_sal.append(emp)

print("High Sal: ", high_sal)
print("Low Sal: ", low_sal)

with open('/content/sample_data/high_salary.csv', 'w') as file:
  file.write(csv_format)
  for i in high_sal:
    file.write(f"{",".join(i)}\n")

with open('/content/sample_data/low_salary.csv', 'w') as file:
  file.write(csv_format)
  for i in low_sal:
    file.write(f"{",".join(i)}\n")
```

3. Develop a program for log file processing:
- Read a log file containing timestamps and events
- Create summary files:
  - errors.txt (all error messages)
  - warnings.txt (all warning messages)
  - summary.txt (count of each type of event)
  - Original format: [timestamp] [type] [message]

```python
errors = []
warnings = []
infos = []
with open('/content/sample_data/application.log') as file:
  for line in file:
    event_type = line.strip("\n ").split()[2]
    if event_type == '[INFO]':
      infos.append(line)
    elif event_type == '[WARNING]':
      warnings.append(line)
    elif event_type == '[ERROR]':
      errors.append(line)

with open('/content/sample_data/errors.txt', 'w') as file:
  for line in errors:
    file.write(line)
with open('/content/sample_data/warnings.txt', 'w') as file:
  for line in warnings:
    file.write(line)

count_errors = len(errors)
count_warnings = len(warnings)
count_infos = len(infos)

summary = f"""
Event Summary Report
--------------------
INFO: {count_infos}
WARNING: {count_warnings}
ERROR: {count_errors}
TOTAL EVENTS: {count_errors + count_warnings + count_infos}
"""

with open('/content/sample_data/summary.txt', 'w') as file:
  file.write(summary.strip())

print(summary.strip())
```

```python
event_counts = {}
event_lines = {}

with open('/content/sample_data/application.log') as file:
    for line in file:
        parts = line.split('] ')
        if len(parts) < 2:
            continue

        event_type = parts[1].strip()

        event_counts[event_type] = event_counts.get(event_type, 0) + 1
        event_lines.setdefault(event_type, []).append(line)

# Write separate files
for event_type, lines in event_lines.items():
    filename = event_type.strip('[]').lower() + "s.txt"
    with open(f'/content/sample_data/{filename}', 'w') as file:
        file.writelines(lines)

# Write summary
total = sum(event_counts.values())

summary_lines = ["Event Summary Report", "--------------------"]

for event_type, count in event_counts.items():
    summary_lines.append(f"{event_type.strip('[]')}: {count}")

summary_lines.append(f"TOTAL EVENTS: {total}")

with open('/content/sample_data/summary.txt', 'w') as file:
    file.write("\n".join(summary_lines))

print("\n".join(summary_lines))
```

4. Create a program for student attendance tracking:
- Read daily attendance files (date_attendance.csv)
- Calculate monthly attendance
- Generate report.csv with columns:
  - Student ID
  - Total Present Days
  - Total Absent Days
  - Attendance Percentage
  - Handle missing files and incomplete data

```python
import glob

file_paths = glob.glob('/content/sample_data/Attendance/*')
student_attendance = {}

for path in file_paths:

  with open(path) as file:
    csv_format = file.readline()
    for line in file:
      student = line.strip().split(',')
      student_name = student[0]
      student_status = student[1]
      if student_name == '' or student_status == '':
        continue
      elif student_status == 'Present':
        if student_attendance.get(student_name) != None:
          student_attendance[student_name]['Present'] += 1
        else:
          student_attendance[student_name] = {'Present': 1, 'Absent': 0}
      elif student_status == 'Absent':
        if student_attendance.get(student_name) != None:
          student_attendance[student_name]['Absent'] += 1
        else:
          student_attendance[student_name] = {'Present': 0, 'Absent': 1}

print(student_attendance)

with open('/content/sample_data/report.txt', 'w') as file:
  # CSV Format
  file.write('StudentID,Total Present Days,Total Absent Days,Attendance Percentage\n')
  for student, attendance in student_attendance.items():
    # Calculate percentage
    percentage = attendance['Present'] / (attendance['Present'] + attendance['Absent']) * 100
    file.write(f"{student},{attendance['Present']},{attendance['Absent']},{percentage}\n")
```

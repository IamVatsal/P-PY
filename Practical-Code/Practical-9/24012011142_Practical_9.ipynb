```python
# Que 1
import mysql.connector

config = {
    'user' : 'root',
    'password' : 'vatsal',
    'host' : 'localhost',
}

try:
    cnx = mysql.connector.connect(**config)
    if cnx.is_connected():
        print("Connected to MySQL database")
except mysql.connector.Error as err:
    print(f"Error: {err}")

cursor = cnx.cursor()

query = "CREATE DATABASE IF NOT EXISTS demo_data"
cursor.execute(query)
cnx.commit()
query = "SHOW DATABASES"
cursor.execute(query)
results = cursor.fetchall()
print("Databases:")
for row in results:
    print(f"Name: {row[0]}")

query = "USE demo_data"
cursor.execute(query)
```

```python
# Que 7

import sqlite3

cnx = sqlite3.connect(':memory:')
cursor = cnx.cursor()

query = "DROP TABLE IF EXISTS student_details"
cursor.execute(query)
cnx.commit()

query = """
CREATE TABLE IF NOT EXISTS student_details (
    id INTEGER PRIMARY KEY AUTOINCREMENT, 
    student_name TEXT, 
    age INTEGER,
    course TEXT
)
"""
cursor.execute(query)

query = """
INSERT INTO student_details (student_name, age, course) VALUES (?, ?, ?)
"""
data = [
    ("Vatsal", 21, "Computer Engineering"),
    ("Honey", 22, "Information Technology"),
    ("Aryan", 20, "MBBS"),
]
cursor.executemany(query, data)
cnx.commit()

query = "SELECT * FROM student_details"
cursor.execute(query)
results = cursor.fetchall()
print("Student Details:")
for row in results:
    print(f"ID: {row[0]}, Name: {row[1]}, Age: {row[2]}, Course: {row[3]}")
```
